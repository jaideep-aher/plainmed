# PlainMed Evaluation Exploration

This notebook loads the saved evaluation CSV and chart for lightweight inspection. Nothing in the application depends on this notebook.

```python
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

results_path = Path("../data/outputs/eval_results.csv")
chart_path = Path("../data/outputs/readability_chart.png")

results = pd.read_csv(results_path)
results.head()
```

```python
summary = results[["fk_original", "fk_base", "fk_finetuned"]].mean().round(2)
summary.to_frame(name="mean_fk_grade")
```
